```python
import os
import time
import requests
from pathlib import Path
from dotenv import load_dotenv
from langchain_google_genai import ChatGoogleGenerativeAI
from langchain_groq import ChatGroq
from langchain_core.tools import tool
from langchain_core.messages import HumanMessage
```

```python
env_path=Path.cwd().parent /".env"
load_dotenv(env_path)
api_key=os.getenv("GEMINI_API_KEY")
api_key2=os.getenv("GROQ_API_KEY")


if not api_key and  api_key2:
    print("API KEY MISSING")
    exit()
    
else:
    print("Initializing Model.....")
```

```python
@tool
def multiply(a:int,b:int) -> int:
    """Multiply two numbers from given two numbers"""
    return a*b
```

```python
print(multiply.invoke({'a':3,'b':5}))
```

```python
# tool binding

llm2=ChatGroq(model='openai/gpt-oss-20b',api_key=api_key2,max_tokens=900)
llm=ChatGoogleGenerativeAI(model='gemini-3.8-flash',api_key=api_key)
```

```python
llm_with_tools=llm.bind_tools([multiply])
```

```python
llm_with_tools.invoke('HI how are you')
```

```python
query=HumanMessage('can you multiply 3 with 10000')
```
